# ***Prompt Template Text Summarization***

```python
# Import necessary classes for creating a prompt template and LLM chain
from langchain.chains import LLMChain
from langchain import PromptTemplate
```

```python
# Define a generic template for summarization and translation
generictemplate = """
Write a summary of the following speech:
Speech:{speech}
Translate the precise summary to {language}
"""
```

```python
# Create a PromptTemplate instance with the defined template
prompt = PromptTemplate(
    input_variables=['speech', 'language'],
    template=generictemplate
)
```

```python
prompt
```

```python
# Format the prompt with the speech text and target language
complete_prompt = prompt.format(speech=speech, language="French")
```

```python
# Display the complete prompt
complete_prompt
```

```python
# Get the number of tokens in the complete prompt
llm.get_num_tokens(complete_prompt)
```

```python
# Create an LLMChain instance with the LLM and prompt template
llm_chain = LLMChain(llm=llm, prompt=prompt)
```

```python
# Run the LLMChain to get the summary in English
summary = llm_chain.run({'speech': speech, 'language': 'enlgish'})
```

```python
summary
```

```python
print(summary)
```

```python
# Run the LLMChain to get the summary in Hindi
summary = llm_chain.run({'speech': speech, 'language': 'hindi'})
```

# ***StuffDocumentChain Text summarization***

```python
# Import the PyPDFLoader class for loading PDF documents
from langchain_community.document_loaders import PyPDFLoader
```

```python
# Load a PDF document using PyPDFLoader
loader = PyPDFLoader("apjspeech.pdf")
```

```python
# Load and split the document into individual pages
docs = loader.load_and_split()
```

```python
docs
```

```python
# Define a template for summarization
template = """ Write a concise and short summary of the following speech,
Speech :{text}"""
```

```python
# Create a PromptTemplate instance with the defined template
prompt = PromptTemplate(input_variables=['text'], template=template)
```

```python
# Import the load_summarize_chain function for creating summarization chains
from langchain.chains.summarize import load_summarize_chain
```

```python
# Create a summarization chain using the 'stuff' method
chain = load_summarize_chain(llm, chain_type='stuff', prompt=prompt, verbose=True)
```

```python
# Run the summarization chain on the loaded documents
output_summary = chain.run(docs)
```

```python
output_summary
```

```python
print(output_summary)
```

# ***Map reduce to Summarize Large documents***

```python
# Import the RecursiveCharacterTextSplitter class for splitting text into chunks
from langchain.text_splitter import RecursiveCharacterTextSplitter
```

```python
docs
```

```python
# Split the documents into smaller chunks using RecursiveCharacterTextSplitter
final_documents = RecursiveCharacterTextSplitter(chunk_size=2000, chunk_overlap=100).split_documents(docs)
```

```python
final_documents
```

```python
len(final_documents)
```

```python
# Define a template for summarizing chunks of text
chunks_prompt = """
Please summarize the below speech:
Speech:`{text}'
Summary:
"""
```

```python
# Create a PromptTemplate instance for summarizing chunks
map_prompt_template = PromptTemplate(input_variables=['text'], template=chunks_prompt)
```

```python
# Define a final prompt template for combining summaries
final_prompt = '''
Provide the final summary of the entire speech with these important points.
Add a Motivation Title, Start the precise summary with an introduction and provide the summary in number 
points for the speech.
Speech:{text}
'''
```

```python
# Create a PromptTemplate instance for the final summary
final_prompt_template = PromptTemplate(input_variables=['text'], template=final_prompt)
```

```python
# Display the final prompt template
final_prompt_template
```

```python
# Create a summarization chain using the 'map_reduce' method
summary_chain = load_summarize_chain(
    llm=llm,
    chain_type="map_reduce",
    map_prompt=map_prompt_template,
    combine_prompt=final_prompt_template,
    verbose=True
)
```

```python
# Run the summarization chain on the final documents
output = summary_chain.run(final_documents)
```

```python
output
```

```python
print(output)
```

# ***Refine Chain For Summarization***

```python
# Create a summarization chain using the 'refine' method
chain = load_summarize_chain(
    llm=llm,
    chain_type="refine",
    verbose=True
)
```
