```python
model_id1 = "dreamlike-art/dreamlike-diffusion-1.0"
```

```python
pipe = StableDiffusionPipeline.from_pretrained(model_id1, torch_dtype=torch.float16, use_safetensors=True)
pipe = pipe.to("cuda")
```

```python
prompt = "A serene Japanese garden with cherry blossoms in full bloom, a red bridge over a koi pond, soft sunlight filtering through the trees."
```

### ***Working with Stable Diffusion parameters***

* Negative prompting
* num_inference_steps
* height
* weight
* num_images_per_prompt

```python
def generate_image(pipe, prompt, params):
  img = pipe(prompt, **params).images

  num_images = len(img)
  if num_images>1:
    fig, ax = plt.subplots(nrows=1, ncols=num_images)
    for i in range(num_images):
      ax[i].imshow(img[i]);
      ax[i].axis('off');

  else:
    fig = plt.figure()
    plt.imshow(img[0]);
    plt.axis('off');
  plt.tight_layout()
```

```python
prompt = "A digital painting of an ancient library where books float in midair, illuminated by magical orbs of light"
```

```python
params = {}
```

```python
generate_image(pipe, prompt, params)
```

```python
#num inference steps
params = {'num_inference_steps': 100}

generate_image(pipe, prompt, params)
```

```python
# height width
params = {'num_inference_steps': 100, 'width': 512, 'height': int(1.5*640)}

generate_image(pipe, prompt, params)
```

```python
# num_images_per_prompt
params = {'num_inference_steps': 100, 'num_images_per_prompt': 2}

generate_image(pipe, prompt, params)
```

# ***Text_to_speech_generation***

```python
from transformers import pipeline
```

```python
text = "Python is a high-level, general-purpose programming language."

pipe = pipeline("text-to-speech", model="suno/bark-small", device="cuda")

output = pipe(text)
```
